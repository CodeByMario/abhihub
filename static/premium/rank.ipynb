```python
import pandas as pd

data = pd.read_csv('output.csv', index_col=None)
```

```python
filter_data = data[['author', 'type']]
filter_data.head()
```

```python
def count(df, author, type):
    count = df[(df['author'] == author) & (df['type'] == type)].shape[0]
    return count

bkp = count(filter_data, "Bhushan Khairkar", "Papers")
bkp
```

```python
df_unique = data['author'].unique()
type = data['type'].unique()
```

```python
type
```

```python
dic = {}
for author in df_unique:
    if author not in dic:
        dic[author] = []
    for i in type:
        if i == 'Notes' or i == 'NOTES':
            dic[author].append(count(filter_data, author, i)*3)
        else:
            dic[author].append(count(filter_data, author, i))
        
```

```python
points = {}
for auth in dic:
    points[auth] = sum(dic[auth])
```
